```python
a=int(input("enter first number: "))
b=int(input("enter second number: "))
while b !=0:
    a, b = b, a % b
print("GCD=",abs(a))
```

```python
n=int(input("enter the number of steps: "))
for i in range(1, n + 1):
    for j in range(1, i + 1):
        print(i * j, end="")
    print()
```

```python
n = 5
for i in range(1, n + 1):
    for j in range(i):
        print("*", end="")
    print()
for i in range(n - 1, 0, -1):
    for j in range(i):
        print("*", end="")
    print()
```

```python
square= lambda a:a*a
rectangle=lambda l,b:l*b
triangle=lambda b,h:0.5*b*h
print("Area of square:",square(5))
print("Area of rectangle:",rectangle(4,6))
print("Area of triangle:",triangle(5,4))
```

```python
list1=list(map(int,input("enter first list: ").split()))
list2=list(map(int,input("enter second list: ").split()))
if len(list1) == len(list2):
           print("Lists are of same length")
else:
           print("Lists are not of same length")
           
if sum(list1)==sum(list2):
           print("Lists have same sum")
else:
           print("List do not have same sum")

if set(list1) & set(list2):
           print("common vales exist")
else:
           print("no common values")
```
